```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

# Define a small epsilon around zero to compute the gradient
d = torch.tensor([0.0], requires_grad=True)

delta = 1e-4

# Define the function
def diag_fn(d):
    # Compute diag as defined
    diag = d + torch.where(d >= 0, 1.0, -1.0) * delta
    return diag

# Compute the output
output = diag_fn(d)

# Backpropagate
output.backward()

# Print the derivative at d=0
print(f"Derivative at d={d.item()}:", d.grad.item())

# Prepare input values
d_vals = torch.linspace(-0.002, 0.002, steps=1000, requires_grad=True)

# Compute output
diag_vals = diag_fn(d_vals)

# Plot the function
plt.plot(d_vals.detach().numpy(), diag_vals.detach().numpy())
plt.xlabel('d')
plt.ylabel('diag(d)')
plt.grid(True)
plt.show()

```
